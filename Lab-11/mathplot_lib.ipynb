```python
data = { "Salary" : [25000, 30000, 37000, 28000, 39000, 48000, 55000, 52000, 35000, 26000, 27000, 31000, 33000, 29000, 40000, 36000, 34000, 45000, 47000, 50000]}
```

```python
import pandas as pd
df=pd.DataFrame(data)
```

```python
plt.plot(df["Salary"],color="red",linestyle="-", marker="*",linewidth="1.2")
plt.show()
```

```python
#Histogram
plt.hist(df['Salary'], bins = 5)
plt.show
```

```python
plt.boxplot(df['Salary'])
plt.show()
```

```python
df.loc[20]=0
```

```python
df["dept"] = ['HR', 'IT', 'Finance', 'HR', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR']*2
```

```python
df.tail(10)
```

```python
df['dept'].value_counts()
```

```python
count=df['dept'].value_counts()
```

```python
plt.pie(count, labels = count.index)
plt.axis("equal")
plt.show()
```

```python
plt.pie(count, labels = count.index,autopct="%1.2f")
plt.axis("equal")
plt.show()
```

```python
plt.pie(count, labels = count.index, autopct="%1.2f", explode = [0, 0.1, 0.2])
plt.axis("equal")
plt.show()
```
